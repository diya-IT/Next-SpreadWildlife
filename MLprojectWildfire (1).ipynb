```python
import tensorflow as tf
import numpy as np

# Features we will use as input
input_features = [
    "tmmn",
    "NDVI",
    "elevation",
    "population",
    "pdsi",
    "vs",
    "pr",
    "tmmx",
    "sph",
    "th",
    "erc",
    "PrevFireMask"
]

target_feature = "FireMask"


def parse_tfrecord(raw_record):

    example = tf.train.Example()
    example.ParseFromString(raw_record.numpy())

    X = []

    for feature_name in input_features:

        values = example.features.feature[
            feature_name
        ].float_list.value

        values = np.array(values, dtype=np.float32)

        values = values.reshape(64, 64)

        X.append(values)

    X = np.stack(X, axis=-1)

    y = np.array(
        example.features.feature[
            target_feature
        ].float_list.value,
        dtype=np.float32
    )

    y = y.reshape(64, 64)

    return X, y
```

```python
input_features = [
    "tmmn",
    "NDVI",
    "elevation",
    "population",
    "pdsi",
    "vs",
    "pr",
    "tmmx",
    "sph",
    "th",
    "erc",
    "PrevFireMask"
]

print("Number of input features:", len(input_features))
```

```python
def parse_record(raw_record):

    example = tf.train.Example()
    example.ParseFromString(raw_record.numpy())

    X = []

    for feature in input_features:

        values = example.features.feature[
            feature
        ].float_list.value

        values = np.array(
            values,
            dtype=np.float32
        )

        values = values.reshape(64, 64)

        X.append(values)

    # 12 × 64 × 64
    X = np.array(X)

    # Change to 64 × 64 × 12
    X = np.transpose(X, (1, 2, 0))

    # Target
    y = np.array(
        example.features.feature[
            "FireMask"
        ].float_list.value,
        dtype=np.float32
    )

    y = y.reshape(64, 64, 1)

    return X, y
```

```python
cnn_model = models.Sequential([

    layers.Input(shape=(64, 64, 12)),

    layers.Conv2D(
        32,
        kernel_size=3,
        padding="same",
        activation="relu"
    ),

    layers.BatchNormalization(),

    layers.Conv2D(
        64,
        kernel_size=3,
        padding="same",
        activation="relu"
    ),

    layers.BatchNormalization(),

    layers.Conv2D(
        64,
        kernel_size=3,
        padding="same",
        activation="relu"
    ),

    layers.BatchNormalization(),

    layers.Conv2D(
        1,
        kernel_size=1,
        padding="same",
        activation="sigmoid"
    )
])

cnn_model.summary()
```

```python
def weighted_binary_crossentropy(y_true, y_pred):

    # Weight for each class
    fire_weight = 10.0
    no_fire_weight = 1.0

    # Avoid log(0)
    y_pred = tf.clip_by_value(
        y_pred,
        1e-7,
        1 - 1e-7
    )

    loss = -(
        no_fire_weight * (1 - y_true) * tf.math.log(1 - y_pred)
        +
        fire_weight * y_true * tf.math.log(y_pred)
    )

    return tf.reduce_mean(loss)
```

```python
cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss=weighted_binary_crossentropy
)

print("CNN compiled successfully!")
```

```python
import tensorflow as tf
import numpy as np

# -----------------------------
# Function to read one record
# -----------------------------
def data_generator(file_list):

    for file_path in file_list:

        dataset = tf.data.TFRecordDataset(file_path)

        for raw_record in dataset:

            X, y = parse_record(raw_record)

            yield X.astype(np.float32), y.astype(np.float32)


# -----------------------------
# Train / Validation files
# -----------------------------

train_files = [
    f for f in tfrecord_files
    if "train" in f
]

eval_files = [
    f for f in tfrecord_files
    if "eval" in f
]

print("Training files:", len(train_files))
print("Validation files:", len(eval_files))


# -----------------------------
# Create TensorFlow datasets
# -----------------------------

train_dataset = tf.data.Dataset.from_generator(
    lambda: data_generator(train_files),
    output_signature=(
        tf.TensorSpec(
            shape=(64, 64, 12),
            dtype=tf.float32
        ),
        tf.TensorSpec(
            shape=(64, 64, 1),
            dtype=tf.float32
        )
    )
)

eval_dataset = tf.data.Dataset.from_generator(
    lambda: data_generator(eval_files),
    output_signature=(
        tf.TensorSpec(
            shape=(64, 64, 12),
            dtype=tf.float32
        ),
        tf.TensorSpec(
            shape=(64, 64, 1),
            dtype=tf.float32
        )
    )
)

# Shuffle training data
train_dataset = train_dataset.shuffle(
    buffer_size=500
)

# Batch
train_dataset = train_dataset.batch(16)
eval_dataset = eval_dataset.batch(16)

# Prefetch
train_dataset = train_dataset.prefetch(
    tf.data.AUTOTUNE
)

eval_dataset = eval_dataset.prefetch(
    tf.data.AUTOTUNE
)

print("Datasets created successfully!")
```

```python
# Calculate mean and standard deviation
# using the first 1000 training samples

channel_sum = np.zeros(12, dtype=np.float64)
channel_squared_sum = np.zeros(12, dtype=np.float64)

total_pixels = 0

for X, y in data_generator(train_files):

    # X shape = (64, 64, 12)

    X = X.astype(np.float64)

    # Sum over height and width
    channel_sum += np.sum(X, axis=(0, 1))

    channel_squared_sum += np.sum(
        X ** 2,
        axis=(0, 1)
    )

    total_pixels += 64 * 64

    if total_pixels >= 1000 * 64 * 64:
        break


# Mean
channel_mean = channel_sum / total_pixels

# Variance
channel_variance = (
    channel_squared_sum / total_pixels
    - channel_mean ** 2
)

# Standard deviation
channel_std = np.sqrt(
    np.maximum(channel_variance, 1e-8)
)

print("Feature statistics:\n")

for i, feature in enumerate(input_features):

    print(
        f"{feature:15s} "
        f"mean = {channel_mean[i]:.4f}, "
        f"std = {channel_std[i]:.4f}"
    )
```

```python
def normalized_generator(file_list):

    for X, y in data_generator(file_list):

        X = (
            X - channel_mean
        ) / channel_std

        yield (
            X.astype(np.float32),
            y.astype(np.float32)
        )


train_dataset = tf.data.Dataset.from_generator(
    lambda: normalized_generator(train_files),
    output_signature=(
        tf.TensorSpec(
            shape=(64, 64, 12),
            dtype=tf.float32
        ),
        tf.TensorSpec(
            shape=(64, 64, 1),
            dtype=tf.float32
        )
    )
)

eval_dataset = tf.data.Dataset.from_generator(
    lambda: normalized_generator(eval_files),
    output_signature=(
        tf.TensorSpec(
            shape=(64, 64, 12),
            dtype=tf.float32
        ),
        tf.TensorSpec(
            shape=(64, 64, 1),
            dtype=tf.float32
        )
    )
)

train_dataset = (
    train_dataset
    .shuffle(500)
    .batch(16)
    .prefetch(tf.data.AUTOTUNE)
)

eval_dataset = (
    eval_dataset
    .batch(16)
    .prefetch(tf.data.AUTOTUNE)
)

print("Normalized datasets ready!")
```

```python
cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss=weighted_binary_crossentropy,
    metrics=[
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)

print("CNN compiled successfully!")
```
